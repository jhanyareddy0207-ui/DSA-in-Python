### Q3. Strings
Write a Python program to accept a string and: Count the number of characters, Count vowels and consonants, Reverse the string, Check whether it is a palindrome.

```python
st = input("Give a string: ")
print("Character count =", len(st))

vow = sum(1 for c in st if c.lower() in "aeiou")
cons = sum(1 for c in st if c.isalpha() and c.lower() not in "aeiou")

print("Vowel count =", vow)
print("Consonant count =", cons)

rev = "".join(reversed(st))
print("Reversed string =", rev)

if st.lower() == rev.lower():
    print("Palindrome: Yes")
else:
    print("Palindrome: No")
```

### Q4. Sets and Dictionaries
Create two sets of integers and demonstrate union, intersection, difference, and symmetric difference. Also create a dictionary containing the names and marks of five students and display the student with the highest marks.

```python
s_a = {2, 4, 6, 8, 10}
s_b = {6, 8, 10, 12, 14}

print("Set 1:", s_a)
print("Set 2:", s_b)
print("Union:", s_a | s_b)
print("Intersection:", s_a & s_b)
print("Difference (s_a - s_b):", s_a - s_b)
print("Symmetric Difference:", s_a ^ s_b)

students = {"Priya": 88, "Arjun": 94, "Deepak": 79, "Simran": 96, "Kabir": 85}

highest_student = ""
max_score = -1
for name, score in students.items():
    if score > max_score:
        max_score = score
        highest_student = name

print("Top performer:", highest_student, "with score", max_score)
```

### Q5. GPA Calculator
Develop a Python program that accepts the grade/grade point and credit of different courses and calculates the student's SGPA/GPA using: GPA = Σ(Grade Point × Credit) / ΣCredits

```python
num_courses = int(input("Enter number of courses: "))
pts = []
crd = []

for i in range(num_courses):
    pts.append(float(input(f"Grade point of course {i+1}: ")))
    crd.append(float(input(f"Credits for course {i+1}: ")))

total_weighted = sum(p * c for p, c in zip(pts, crd))
total_crd = sum(crd)

gpa = total_weighted / total_crd
print("SGPA:", round(gpa, 2))
```

### Q6. Functions and Iteration
Write separate Python functions to: Calculate the factorial of a number, Generate the Fibonacci series up to n terms, Check whether a number is prime.

```python
def get_factorial(n):
    ans = 1
    for i in range(2, n + 1):
        ans *= i
    return ans

def get_fibonacci(n):
    if n <= 0:
        return []
    seq = [0, 1]
    while len(seq) < n:
        seq.append(seq[-1] + seq[-2])
    return seq[:n]

def check_prime(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

print("5! =", get_factorial(5))
print("Fibonacci series up to 10 terms:", get_fibonacci(10))
print("Check prime 13:", check_prime(13))
print("Check prime 15:", check_prime(15))
```

### Q9. List Comprehension
Using list comprehension, generate: Squares of numbers from 1 to 20, Even numbers from 1 to 50, Numbers divisible by both 3 and 5 from 1 to 100.

```python
sq_nums = [n**2 for n in range(1, 21)]
print("Squares:", sq_nums)

even_nums = [n for n in range(1, 51) if n % 2 == 0]
print("Even numbers:", even_nums)

div_nums = [n for n in range(1, 101) if n % 3 == 0 and n % 5 == 0]
print("Divisible by 3 and 5:", div_nums)
```

### Q10. Mini Programming Problem
Develop a menu-driven Python program for maintaining a student record using a dictionary. The program should provide options to: Add a student, Update marks, Search for a student, Delete a student, Display all student records.

```python
db = {}

while True:
    print("\nMenu: [1]Add [2]Update [3]Search [4]Delete [5]Display [6]Exit")
    op = input("Choose: ").strip()

    if op == "1":
        name = input("Student name: ")
        marks = float(input("Marks: "))
        db[name] = marks
        print("Saved.")
    elif op == "2":
        name = input("Student name: ")
        if name in db:
            db[name] = float(input("Updated marks: "))
            print("Updated successfully.")
        else:
            print("Not found.")
    elif op == "3":
        name = input("Student name: ")
        if name in db:
            print("Marks of", name, "=", db[name])
        else:
            print("No record found.")
    elif op == "4":
        name = input("Student name: ")
        if name in db:
            db.pop(name)
            print("Record removed.")
        else:
            print("Student not present.")
    elif op == "5":
        if not db:
            print("Empty records.")
        else:
            for k, v in db.items():
                print(f"{k} : {v}")
    elif op == "6":
        print("Done.")
        break
    else:
        print("Invalid option!")
```
